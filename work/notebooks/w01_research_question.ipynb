## 3. Quick look at the data (2-3 real numbers)

The starter dataset contains 30,000 pages. Of these, 6,068 pages have at least 100 impressions but 0 clicks, and 1,347 of those pages are in the page_1 position tier. This shows that there are measurable search-performance signals that can help prioritize pages for human review.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

url = "https://raw.githubusercontent.com/Lokeshtiwari723/Proto-ex/main/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(url)

zero_click = df[
    (df["impressions_90d"] >= 100) &
    (df["clicks_90d"] == 0)
]

print("Total pages:", len(df))
print("Pages with 100+ impressions but 0 clicks:", len(zero_click))
print("Pages in page_1 tier within that group:", (zero_click["position_tier"] == "page_1").sum())
```
