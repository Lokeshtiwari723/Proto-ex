## 5. Why ML beats a fixed rule here

A fixed rule could prioritize pages using one or two hand-set thresholds, such as high impressions with zero clicks. That is easy to explain but can miss useful combinations of signals and requires manual threshold tuning. ML is useful here because it can combine multiple observable signals into a consistent ranking that can be evaluated with Precision@K. The model would support human review rather than replace it or prove causality.
